## Data Visualization with Seaborn
Seaborn is a powerful Python library for data visualization that is built on top of Matplotlib. It provides a high-level interface for creating attractive and informative statistical graphics. In this guide, we will explore some of the key features of Seaborn and how to use it effectively.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Basic Plotting with Seaborn

tips=sns.load_dataset('tips')
tips.head(5)
```

```python
# Create a scatter plot
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
# Line Plot

sns.lineplot(x='size', y='total_bill', data=tips)
plt.title('Line Plot of Total Bill vs Size')
plt.xlabel('Size')
plt.ylabel('Total Bill')
plt.show()
```

```python
# Catergorical Plotting with Seaborn
# Bar plot

sns.barplot(x='day', y='total_bill', data=tips)
plt.title('Bar Plot of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
# Box plot

sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Box Plot of Total Bill by Day')
```

```python
# Violin plot

sns.violinplot(x='day', y='total_bill', data=tips)
```

```python
# Histogram
sns.histplot(tips['total_bill'], bins=10, kde=True)
```

```python
# Kde plot
sns.kdeplot(tips['total_bill'], shade=True)
```

```python
# Pairplot

sns.pairplot(tips)
```

```python
# Heatmap
corr_matrix = tips[['total_bill', 'tip', 'size']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
```
