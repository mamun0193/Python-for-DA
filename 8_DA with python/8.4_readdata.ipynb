### Reading Data from different Sources

When working with data, it's important to know how to read data from various sources. Here are some common methods for reading data:
1. **Reading from CSV Files**: 
   - You can use libraries like `pandas` in Python to read CSV files easily.
   ```python
   import pandas as pd
   data = pd.read_csv('data.csv')
   ```
2. **Reading from Excel Files**:
    - The `pandas` library also provides functionality to read Excel files.
    ```python
    data = pd.read_excel('data.xlsx')
    ```
3. **Reading from SQL Databases**:
    - You can use libraries like `SQLAlchemy` or `sqlite3` to connect to SQL databases and read data.
    ```python
    import sqlite3
    conn = sqlite3.connect('database.db')
    data = pd.read_sql_query('SELECT * FROM table_name', conn)
    ```
4. **Reading from JSON Files**:
    - JSON files can be read using the `json` library or `pandas`.
    ```python
    import json
    with open('data.json') as f:
        data = json.load(f)
    ```
    - Alternatively, using `pandas`:
    ```python
    data = pd.read_json('data.json')
    ```
5. **Reading from APIs**:
    - You can use the `requests` library to fetch data from APIs.
    ```python
    import requests
    response = requests.get('https://api.example.com/data')
    data = response.json()
    ```
With all these methods, you can easily read data from different sources and start analyzing it. Make sure to handle exceptions and errors appropriately when reading data to ensure your program runs smoothly.
