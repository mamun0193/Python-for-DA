## Pandas-DataFrame and Series
Pandas is a powerful data manipulation library in Python that provides two primary data structures: DataFrame and Series.It allows for efficient handling of structured data, making it easier to analyze and manipulate datasets.

### Series
A Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). It is similar to a column in a spreadsheet or a SQL table. Each element in a Series has an associated index, which allows for easy access and manipulation of data.
#### Creating a Series
```python
import pandas as pd
data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print(series)
```
##### Accessing Data in Series
You can access data in a Series using indexing and slicing.
```python
# Accessing an element by index
element = series[2]
print(element)
```
```python
# Slicing the Series
subset = series[1:4]
print(subset)
```

### DataFrame
A DataFrame is a two-dimensional labeled data structure with columns of potentially different types. It can be thought of as a table or a spreadsheet, where each column can hold different types of data (e.g., integers, strings, floats). DataFrames are widely used for data analysis and manipulation tasks.
#### Creating a DataFrame
```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}
df = pd.DataFrame(data)
print(df)
```
#### Accessing Data in DataFrame
You can access data in a DataFrame using various methods, such as column selection, row selection, and conditional filtering.
```python
# Accessing a column
ages = df['Age']
print(ages)
```
```python
# Accessing a row by index
row = df.loc[1]
print(row)
```
```python
# Conditional filtering
adults = df[df['Age'] > 30]
print(adults)
```
