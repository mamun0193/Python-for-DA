## Data Manipulation Techniques

Data manipulation techniques are essential for preparing and transforming data for analysis. These techniques can help clean, organize, and structure data to make it more suitable for analysis. Below are some common data manipulation techniques:

### 1. Data Cleaning
Data cleaning involves identifying and correcting errors or inconsistencies in the data. This may include handling missing values, removing duplicates, and correcting data types. Common methods for data cleaning include:
- **Handling Missing Values**: Techniques such as imputation, deletion, or using algorithms that can handle missing data.
- **Removing Duplicates**: Identifying and removing duplicate records to ensure data integrity.
- **Correcting Data Types**: Ensuring that data is in the correct format (e.g., converting strings to dates or numbers).
### 2. Data Transformation
Data transformation involves changing the format, structure, or values of data to make it more suitable for analysis. This can include normalization, scaling, and encoding categorical variables. Common methods for data transformation include:
- **Normalization**: Adjusting the scale of data to a standard range, often between 0 and 1.
- **Scaling**: Rescaling data to have a specific mean and standard deviation, often used in machine learning algorithms.
- **Encoding Categorical Variables**: Converting categorical data into numerical format using techniques like one-hot encoding or label encoding.
### 3. Data Aggregation
Data aggregation involves summarizing data to provide insights or to prepare it for analysis. This can include grouping data by specific attributes and calculating summary statistics. Common methods for data aggregation include:
- **Grouping**: Organizing data into groups based on specific attributes (e.g., grouping sales data by region).
- **Calculating Summary Statistics**: Computing measures such as mean, median, sum, and count for each group to gain insights into the data.
### 4. Data Merging and Joining
Data merging and joining involve combining multiple datasets into a single dataset based on common attributes. This is useful for integrating data from different sources. Common methods for data merging and joining include:
- **Inner Join**: Combining datasets where only matching records from both datasets are included.
- **Outer Join**: Combining datasets where all records from both datasets are included, with missing values filled in where there is no match.
- **Left Join**: Combining datasets where all records from the left dataset are included, and matching records from the right dataset are added.
- **Right Join**: Combining datasets where all records from the right dataset are included, and matching records from the left dataset are added.



```python
import pandas as pd
```

```python
df=pd.read_csv('sd.csv')

## fetch the first 5 rows of the data
df.head(5)

## Fetch the last 5 rows of the data
df.tail(5)
```

```python
df.describe()
```

```python
df.dtypes # It will show the data type of each column in the dataset
```

```python
# Handling missing values

df.isnull()# It will show the columns which have missing values
```

```python
df.isnull().sum() # It will show the total number of missing values in each column
```

```python
df_filled=df.fillna(0) # It will fill the missing values with 0
print(df_filled)
```

```python
## Filling missing value with mean of the column(any desired column can be used)

# We'll use sales column to fill the missing values with mean of the sales column

df['Sales'].fillna(df['Sales'].mean(), inplace=True)
```

```python
## Renaming the columns of the dataset

df.rename(columns={'OrderProfitable':'Profit'}, inplace=True)
df.head(5)
```

```python
## Change Datatype of a column

df['NewSalesperCustomer']= df['SalesperCustomer'].astype('int64') # It will change the datatype of the column to int64
df.head(5)

```

```python
# Data Aggregation and Grouping


grouped_mean= df.groupby('CustomerName')['Discount'].mean() # It will group the data by CustomerName and calculate the mean of the Value column
print(grouped_mean)
```

```python
## Aggregate Multiple Functions on a Column

grouped_agg= df.groupby('CustomerName')['Discount'].agg(['mean','sum','max']) # It will group the data by CustomerName and calculate the mean, sum and max of the Value column
print(grouped_agg)
```

```python
# Merging and Joining DataFrames

# Create a sample DataFrame to demonstrate merging and joining

df1 = pd.DataFrame({
    'CustomerID': [1, 2, 3, 4],
    'CustomerName': ['Alice', 'Bob', 'Charlie', 'David'],
    'Country': ['USA', 'Canada', 'UK', 'Australia']
})
df2 = pd.DataFrame({
    'CustomerID': [3, 4, 5, 6],
    'OrderID': [101, 102, 103, 104],
    'OrderAmount': [250, 300, 150, 400]
})

## Merging DataFrames using inner join

merged_inner = pd.merge(df1, df2, on='CustomerID', how='inner')
print("Inner Join Result:")
print(merged_inner)

## Merging DataFrames using left join

merged_left = pd.merge(df1, df2, on='CustomerID', how='left')
print("Left Join Result:")
print(merged_left)

## Mergeing DataFrames using right join 

merged_right = pd.merge(df1, df2, on='CustomerID', how='right')
print("Right Join Result:")
print(merged_right)

## Merging DataFrames using outer join

merged_outer = pd.merge(df1, df2, on='CustomerID', how='outer')
print("Outer Join Result:")
print(merged_outer)


```
