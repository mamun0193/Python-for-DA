## Descriptive Statistics
Descriptive statistics are used to summarize and describe the main features of a dataset. They provide simple summaries about the sample and the measures.

### Measure of Central Tendency
The measures of central tendency are statistical measures that describe the center point of a dataset. They include:
- **Mean**: Calculated by adding all the values together and dividing by the number of values.
- **Median**: Found by arranging the data in order and selecting the middle value.
- **Mode**: Identified by determining which value appears most frequently in the dataset.

#### Mean
The mean is calculated using the formula:
\[ \text{Mean} = \frac{\sum_{i=1}^{n} x_i}{n} \]
Where \( x_i \) represents each value in the dataset and \( n \) is the total number of values.

**Characteristics of Mean**:
- Sensitive to outliers: Extreme values can significantly affect the mean. e.g ., in the dataset [1, 2, 3, 4, 100], the mean is 22, which does not represent the majority of the data.
- Uses all data points in its calculation. 
- Can be influenced by the scale of the data. 

#### Median
The median is the middle value of a dataset when it is ordered from least to greatest. If there is an even number of observations, the median is the average of the two middle numbers.
The median is calculated using the following steps:
1. Arrange the data in ascending order.
2. If the number of observations (n) is odd, the median is the value at position \( \frac{n+1}{2} \).
3. If the number of observations (n) is even, the median is the average of the values at positions \( \frac{n}{2} \) and \( \frac{n}{2} + 1 \).

**Characteristics of Median**:
- Not affected by outliers: The median provides a better measure of central tendency for skewed distributions. e.g ., in the dataset [1, 2, 3, 4, 100], the median is 3, which better represents the central tendency of the data.
- Represents the 50th percentile of the data.
- Can be used for ordinal data.

#### Mode
The mode is the value that appears most frequently in a dataset. A dataset may have one mode (unimodal), more than one mode (bimodal or multimodal), or no mode at all.
The mode is calculated by identifying the value(s) that occur most frequently in the dataset.
e.g- For the dataset [1, 2, 2, 3, 4], the mode is 2, as it appears twice.

**Characteristics of Mode**:
- Can be used for nominal data.
- Not affected by outliers.
- Can be used to identify the most common value in a dataset.
