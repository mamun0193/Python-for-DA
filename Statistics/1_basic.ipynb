## Statistics

Statistics is a branch of mathematics that deals with the collection, analysis, interpretation, presentation, and organization of data. It provides tools and methods for making sense of complex data sets and drawing meaningful conclusions.

### Data 
Data refers to raw facts and figures or pieces of information that can be collected, measured, and analyzed. It can be qualitative (categorical) or quantitative (numerical) in nature.

example: 
qualitative data: colors, types of fruits
quantitative data: height, weight, age

### Importance of Statistics in field of Data Science

- **Data Exploration and Summarization:** Statistics helps in understanding the underlying patterns and distributions in the data, which is crucial for making informed decisions.
- **Model Building:** Statistical techniques are used to build predictive models that can forecast future trends and behaviors based on historical data.
- **Statistical Inference:** It allows data scientists to make inferences about populations based on sample data, enabling them to draw conclusions and make predictions.
- **Hypothesis Testing:** Statistics provides methods for testing hypotheses and determining the significance of results, which is essential for validating findings and making data-driven decisions.
- **Optimization and Efficiency:** Statistical methods help in optimizing processes and improving efficiency by identifying key factors that influence outcomes and performance.

## Types of Statistics

### Descriptive Statistics
Descriptive statistics involves **summarizing** and **organizing** the data to make it more understandable. It includes measures such as-
1. **Measures of Central Tendency:** Mean, Median, Mode
2. **Measures of Dispersion:** Range, Variance, Standard Deviation
3. **Data Distribution(Visualization):** Graphs, Charts, Histograms, Box Plots
4. **Summary Statistics:** Five Number Summary, Percentiles, Quartiles

### Inferential Statistics
Inferential statistics involves **making predictions** or **inferences** about a population based on a sample of data. It includes techniques such as-
1. **Hypothesis Testing:** t-tests, chi-square tests, ANOVA
2. **Confidence Intervals:** Estimating population parameters based on sample data
3. **Regression Analysis:** Understanding relationships between variables and making predictions


Example: There are 20 statistic class in our college and we've collcted the heights of students in the class.

- **Descriptive question:** What is the average height of students in the class?
- **Inferential question:** Based on the sample of 20 students, can we infer the average height of all students in the college?


## Population Data vs Sample Data

### Population Data:
Population data refers to the complete set of data that includes all members of a defined group. It represents the entire population being studied and is often used in statistical analysis to draw conclusions about the population as a whole.

#### Characteristics of Population Data:
- **Complete Set:** Contains all the observations or measurements of interest.
- **Parameter:** A nuemerical value summarizing the entire population (e.g., population mean, population variance).

#### Example of Population Data:

- The heights of all students in a university.

### Sample Data:
Sample data refers to a subset of the population data that is selected for analysis. It is used to make inferences about the population without having to collect data from every member of the population.

#### Characteristics of Sample Data:
- **Subset:** Represents a portion of the population.
- **Statistic:** A numerical value summarizing the sample (e.g., sample mean, sample variance).
- **Inference:** Used to estimate population parameters.

#### Example of Sample Data:
- The heights of 100 randomly selected students from a university.
