### Importance of Statistics in field of Data Science

- **Data Exploration and Summarization:** Statistics helps in understanding the underlying patterns and distributions in the data, which is crucial for making informed decisions.
- **Model Building:** Statistical techniques are used to build predictive models that can forecast future trends and behaviors based on historical data.
- **Statistical Inference:** It allows data scientists to make inferences about populations based on sample data, enabling them to draw conclusions and make predictions.

## Types of Statistics

### Descriptive Statistics
Descriptive statistics involves **summarizing** and **organizing** the data to make it more understandable. It includes measures such as-
1. **Measures of Central Tendency:** Mean, Median, Mode
2. **Measures of Dispersion:** Range, Variance, Standard Deviation
3. **Data Distribution(Visualization):** Graphs, Charts, Histograms, Box Plots
4. **Summary Statistics:** Five Number Summary, Percentiles, Quartiles

### Inferential Statistics
Inferential statistics involves **making predictions** or **inferences** about a population based on a sample of data. It includes techniques such as-
1. **Hypothesis Testing:** t-tests, chi-square tests, ANOVA
2. **Confidence Intervals:** Estimating population parameters based on sample data
3. **Regression Analysis:** Understanding relationships between variables and making predictions


Example: There are 20 statistic class in our college and we've collcted the heights of students in the class.

- **Descriptive question:** What is the average height of students in the class?
- **Inferential question:** Based on the sample of 20 students, can we infer the average height of all students in the college?


## Population Data vs Sample Data

### Population Data:
Population data refers to the complete set of data that includes all members of a defined group. It represents the entire population being studied and is often used in statistical analysis to draw conclusions about the population as a whole.

#### Characteristics of Population Data:
- **Complete Set:** Contains all the observations or measurements of interest.
- **Parameter:** A nuemerical value summarizing the entire population (e.g., population mean, population variance).

#### Example of Population Data:

- The heights of all students in a university.

### Sample Data:
Sample data refers to a subset of the population data that is selected for analysis. It is used to make inferences about the population without having to collect data from every member of the population.

#### Characteristics of Sample Data:
- **Subset:** Represents a portion of the population.
- **Statistic:** A numerical value summarizing the sample (e.g., sample mean, sample variance).
- **Inference:** Used to estimate population parameters.

#### Example of Sample Data:
- The heights of 100 randomly selected students from a university.

## Types of Sampling 

1. **Probability Sampling:** Each member of the population has a known, non-zero chance of being selected. Examples include simple random sampling, stratified sampling, and cluster sampling.
- **Simple Random Sampling:** Every member of the population has an equal chance of being selected. For example, selecting people randmoly from a list of all students in a university.
- **Systematic Sampling:** Selecting every nth member from a list of the population. For example, selecting every 10th student from a list of all students in a university.
- **Stratified Sampling:** Dividing the population into subgroups (strata) based on certain characteristics and then randomly sampling from each stratum. For example, dividing employees into departments and randomly selecting employees from each department.
- **Cluster Sampling:** Dividing the population into clusters (groups) and then randomly selecting entire clusters for analysis. For example, randomly selecting schools from a district and analyzing all students in those schools.
- **Multistage Sampling:** A combination of different sampling methods, often used in large-scale surveys. For example, first selecting clusters (schools) and then randomly sampling students within those clusters.

2. **Non-Probability Sampling:** Not every member of the population has a known or equal chance of being selected. Examples include convenience sampling, judgmental sampling, and quota sampling.

- **Convenience Sampling:** Selecting individuals who are easiest to reach or access. For example, surveying students in a classroom because they are readily available.
- **Judgmental Sampling:** Selecting individuals based on the researcher's judgment or expertise. For example, choosing experts in a field to participate in a study.
- **Snowball Sampling:** Existing study subjects recruit future subjects from among their acquaintances. For example, asking participants to refer friends or colleagues for a study.
- **Quota Sampling:** Selecting individuals to meet a predetermined quota based on certain characteristics. For example, ensuring a survey includes a specific number of participants from each age group.

### Scales of Measurement of Data

The scale of measurement refers to the nature of information within the values assign to a variable and how they can be categorized, ordered, or quantified. There are four primary scales of measurement:

1. **Nominal Scale:** This scale categorizes data into distinct groups or categories that don't have an intrinsic order. Examples include gender, race, or types of fruits. Nominal data is qualitative and cannot be used for mathematical operations.

**Characteristics of Nominal Scale:**
- Data is categorized based on names, labels, or qualities. e.g - Male, Female
- No inherent order or ranking among the categories. e.g - Red, Blue, Green
- Catagories are mutually exclusive, meaning each data point can belong to only one category. e.g - M/F, Yes/No, etc.


2. **Ordinal Scale:** This scale classifies data into categories that can be ranked or ordered. Examples include rankings (e.g., first, second, third) or levels of satisfaction (e.g., satisfied, neutral, dissatisfied). Ordinal data is also qualitative.

**Characteristics of Ordinal Scale:**
- Data is categorized and ranked based on a specific order or hierarchy. e.g - Low, Medium, High
- The interval between the ranks may not be equal or meaningful. e.g - 1st, 2nd, 3rd, socio economic status (low, middle, high)


3. **Interval Scale:** This scale categorizes data into ordered groups or categories, where the order matters and the differences between the values are meaningful. However, there is no true zero point. Examples include temperature measured in Celsius or Fahrenheit. Interval data is quantitative.

**Characteristics of Interval Scale:**

- Data is ordered with consistent intervals between values. e.g - Temperature in Celsius or Fahrenheit
- Allow for meaningful comparisons of differences between values. e.g - 20°C is 10°C warmer than 10°C
- No true zero point, meaning that zero does not represent the absence of the quantity being measured. e.g - 0°C does not mean no temperature, it is just a point on the scale.



4. **Ratio Scale:** This scale categorizes data into ordered groups or categories, where the order matters, the differences between the values are meaningful, and there is a true zero point. Examples include height, weight, and age. Ratio data is quantitative and allows for meaningful comparisons using ratios.


**Characteristics of Ratio Scale:**
- Data is ordered with consistent intervals between values. e.g - Height, Weight, Age
- Has a true zero point, meaning that zero represents the absence of the quantity being measured. e.g - 0 kg means no weight, 0 years means no age.
- Differnces are measurable and meaningful, allowing for calculations of ratios. e.g - A person weighing 80 kg is twice as heavy as a person weighing 40 kg.
