## Importing modules in Python

- In Python, modules and packages are used to organize code into manageable sections. A module is a single file (or files) that are imported under one import and used. A package is a collection of modules in directories that give a package hierarchy.

### Rules for Importing Modules

1. **Using the `import` statement**: You can import a module using the `import` statement. For example:
   ```python
   import math
   print(math.sqrt(16))  # Output: 4.0
   ```
2. **Using the `from` statement**: You can import specific attributes or functions from a module using the `from` statement. For example:
   ```python
    from math import sqrt
    print(sqrt(16))  # Output: 4.0
3. **Using aliases**: You can give a module or function an alias using the `as` keyword. For example:
    ```python
    import numpy as np
    print(np.array([1, 2, 3]))  # Output: [1 2 3]
    ```
### Best Practices for Importing Modules
- Import modules at the beginning of the file.
- Avoid using wildcard imports (e.g., `from module import *`) as they can lead to confusion and make the code less readable.
- Use absolute imports rather than relative imports to avoid ambiguity.
- Group related imports together and separate them with a blank line for better readability.

````python
import math
math.sqrt(16) # output: 4.0
````

````python
from math import sqrt,pi
print(sqrt(25)) # output: 5.0
print(pi) # output: 3.141592653589793
````

````python
import numpy as np

np.array([1, 2, 3]) # output: array([1, 2, 3])
````

````python
from math import *
print(sin(pi/2)) # output: 1.0
print(cos(0)) # output: 1.0
print(sqrt(9)) # output: 3.0
print(log(100, 10)) # output: 2.0
print(factorial(5)) # output: 120
````

### Custom Modules and Packages
- You can create your own modules and packages by organizing your code into separate files and directories. For example, you can create a module named `mathematics.py` with functions like `add`, `subtract`, etc., and then import it in your main script.\

````python
from package.mathematics import add
print(add(10, 5)) # output: 15
````

````python
from package import mathematics

mathematics.add(10, 5) # output: 15
````
