## Standard Library 
The Python Standard Library is a collection of modules and packages that come bundled with Python. It provides a wide range of functionality, including file I/O, system calls, data serialization, and more.We can import these modules and use their functions in our code.

for example, we can import the `math` module to perform mathematical operations:

```python
import math
print(math.sqrt(16)) # output: 4.0
```


````python
import array
arr= array.array('i',[1,2,3,4,5])
print(arr) # output: array('i', [1, 2, 3, 4, 5])
````

````python
# random
import random
print(random.randint(1, 10)) # output: random integer between 1 and 10
print(random.choice(['apple', 'banana', 'cherry'])) # output: random choice from the list
````

````python
# File and directory operations
import os
print(os.getcwd()) # output: current working directory
print(os.listdir()) # output: list of files and directories in the current directory
os.mkdir('new_directory') # create a new directory
````

````python
# High level operation on files or collections of files

import shutil
# shutil.copy('source.txt', 'destination.txt') # copy a file
# shutil.move('source.txt', 'destination.txt') # move a file

shutil.rmtree('directory_to_delete') # delete a directory and all its contents
````

````python
# Data Serialization

import json

data= {
    'name': 'Namun',
    'age': 22,
    'city': 'CDG'
}
json_str= json.dumps(data) # convert Python object to JSON string
print(json_str) # output: {"name": "Namun", "age": 22, "city": "CDG"}
print(type(json_str)) # output: <class 'str'>

parsed_data= json.loads(json_str) # convert JSON string back to Python object
print(parsed_data) # output: {'name': 'Namun', 'age': 22, 'city': 'CDG'}
print(type(parsed_data)) # output: <class 'dict'>
````

````python
# CSV file

import csv

with open('data.csv', 'w', newline='') as file:
    writer= csv.writer(file)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Namun', 22, 'CDG'])
    writer.writerow(['Alice', 30, 'NYC'])

    with open('data.csv', 'r') as file:
        reader= csv.reader(file)
        for row in reader:
            print(row)
````

````python
# Datetime

import datetime

current_datetime= datetime.datetime.now()

print(current_datetime) # output: current date and time

yesteday= current_datetime - datetime.timedelta(days=1)
print(yesteday) # output: date and time of yesterday
````

````python
# Time
import time

print(time.time()) # output: current time in seconds since the epoch
time.sleep(2) # pause the program for 2 seconds
print(time.time()) # output: current time in seconds since the epoch after 2 seconds
````

````python
## Regular Expressions

import re

pattern= r'\d+' # match one or more digits
text= 'There are 123 apples and 456 oranges.'
matches= re.findall(pattern, text)
print(matches) # output: ['123', '456']
````
