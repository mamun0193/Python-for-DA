## File Operation- Reading and Writing Files
File hendling is a cruicial part of programming. It allows us to read from and write to files, enabling data persistence and manipulation. In this section, we will explore how to perform basic file operations in Python.

### Reading Files
To read a file in Python, you can use the built-in `open()` function. The `open()` function takes two arguments: the name of the file and the mode in which you want to open the file. The most common modes are:
- `'r'`: Read mode (default) - Opens the file for reading.
- `'w'`: Write mode - Opens the file for writing (creates a new file or truncates an existing file).
Here is an example of how to read a file:
```python
with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```
### Writing Files
To write to a file in Python, you can also use the `open()` function with the `'w'` mode. This will create a new file if it does not exist or overwrite the existing file. Here is an example of how to write to a file:
```python
with open('example.txt', 'w') as file:
    file.write("Hello, World!\n")
    file.write("This is a sample text file.\n")
```

#### Appending to Files
If you want to add content to an existing file without overwriting it, you can use the `'a'` mode. This will append the new content to the end of the file. Here is an example:
```python
with open('example.txt', 'a') as file:
    file.write("This line will be appended to the existing content.\n")
    file.write("Appending another line.\n")
```

````python
# Reading a full file 

with open('example.txt', 'r') as file:
    content= file.read()
    print(content)
````

````python
# Reading a file line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())  # Using strip() to remove leading/trailing whitespace
````

````python
# Reading a file --- IGNORE ---

with open('example.txt', 'r') as file:
    lines = file.readlines()
    for line in lines:
        print(line.strip())
````

````python
# Writing to a file(Overwriting existing content)

with open('example.txt', 'w') as file:
    file.write("This is a new line.\n")
    file.write("This will overwrite the existing content.\n")
    # content = file.read()  # This will not work as the file is opened in write mode

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
````

````python
# Write a file(without overwriting existing content)

with open('example.txt', 'a') as file:
    file.write("This line will be appended to the existing content.\n")
    file.write("Appending another line.\n")
with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
````

````python
# Writing a list of lines to a file

lines= ["First line\n", "Second line\n", "Third line\n"]

with open('example.txt', 'a') as file:
    file.writelines(lines) # This writelines() method writes a list of strings to the file without adding newlines automatically.
with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
````

````python
# Binary file operations

## Writing binary data to a file

data= b'\x00\x01\x02\x03\x04\x05'  # Example binary data
with open('example.bin', 'wb') as file:
    file.write(data)

## Reading binary data from a file

with open('example.bin', 'rb') as file:
    binary_content = file.read()
    print(binary_content)
````

````python
# Read a content from a source text file and write it to a destination text file

## Copying a text file 

with open('example.txt', 'r') as source_file:
    content = source_file.read()
with open('destination.txt', 'w') as dest_file:
    dest_file.write(content)
````

````python
# Read a text file and count number of lines, words, and characters

def count_file_content(file_path):
    with open(file_path, 'r') as file:
        content = file.read()
        lines = content.splitlines()
        words = content.split()
        num_lines = len(lines)
        num_words = len(words)
        num_chars = len(content)
    return num_lines, num_words, num_chars
file_path = 'example.txt'
num_lines, num_words, num_chars = count_file_content(file_path)
print(f"Lines: {num_lines}, Words: {num_words}, Characters: {num_chars}")
````

The w+ mode in Python is used to open a file for both writing and reading. If the file already exists, it will be truncated to zero length, effectively erasing its contents. If the file does not exist, a new file will be created. This mode allows you to write data to the file and then read it back without closing and reopening the file.

````python
# Writing and then reading a file 

with open('example.txt', 'w+') as file:
    file.write("This is a test line.\n")
    file.write("This is another test line.\n")
    file.seek(0)  # Move the cursor to the beginning of the file
    content = file.read()
    print(content)
````
